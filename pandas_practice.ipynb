# Pandas Practice

Pandas is an open-source data manipulation and analysis library built on top of NumPy. It provides high-performance, easy-to-use data structures, namely Series (1D) and DataFrame (2D), making it essential for tabular data analysis.

## 1. Importing Pandas

```python
import pandas as pd
import numpy as np # Often used alongside pandas

print("Pandas version:", pd.__version__)
```

## 2. Core Data Structures: Series
A Series is a one-dimensional labeled array capable of holding any data type.

```python
# Create a Series from a list
s1 = pd.Series([10, 20, 30, 40, 50])
print("Basic Series:\n", s1)

# Create a Series with custom index
s2 = pd.Series([10, 20, 30], index=['a', 'b', 'c'])
print("\nSeries with custom index:\n", s2)

# Create a Series from a dictionary
s3 = pd.Series({'apples': 5, 'oranges': 8, 'bananas': 12})
print("\nSeries from dict:\n", s3)

# Accessing elements
print("\nAccess by index label 'b':", s2['b'])
print("Access by positional index 0:", s2.iloc[0])
```

## 3. Core Data Structures: DataFrame
A DataFrame is a 2-dimensional labeled data structure with columns of potentially different types (like a SQL table or Excel spreadsheet).

```python
# Create DataFrame from a dictionary of lists
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [25, 30, 35, 40],
    'City': ['New York', 'Paris', 'London', 'Tokyo']
}
df = pd.DataFrame(data)
print("DataFrame from dict:\n", df)

# Create DataFrame from a list of dictionaries
data2 = [
    {'Name': 'Eve', 'Age': 22},
    {'Name': 'Frank', 'City': 'Berlin'}
]
df2 = pd.DataFrame(data2)
print("\nDataFrame from list of dicts (handles missing data as NaN):\n", df2)
```

## 4. Reading and Writing Data
Pandas can read from and write to various file formats like CSV, Excel, JSON, SQL, etc.

```python
# Create a dummy CSV file to read
df.to_csv('dummy_data.csv', index=False)

# Read from CSV
df_csv = pd.read_csv('dummy_data.csv')
print("Data read from CSV:\n", df_csv)
```

## 5. Viewing and Inspecting Data
Methods to quickly understand the dataset's structure and contents.

```python
# Create a larger dummy DataFrame for demonstration
np.random.seed(42)
large_df = pd.DataFrame({
    'A': np.random.randn(100),
    'B': np.random.randint(1, 100, 100),
    'Category': np.random.choice(['X', 'Y', 'Z'], 100)
})

# Display first n rows
print("Head (first 3 rows):\n", large_df.head(3))

# Summary statistics for numerical columns
print("\nDescribe (statistics):\n", large_df.describe())
```

## 6. Selection and Filtering
Extracting specific subsets of data.

```python
# Selecting multiple columns (returns a DataFrame)
print("\nSelecting 'A' and 'Category':\n", large_df[['A', 'Category']].head(3))

# Boolean indexing (Filtering)
filtered_df = large_df[large_df['B'] > 80]
print("\nRows where B > 80 (first 3):\n", filtered_df.head(3))
```

## 7. Data Manipulation & Cleaning

```python
# Handling Missing Values
df_missing = pd.DataFrame({'A': [1, 2, np.nan, 4], 'B': [5, np.nan, np.nan, 8]})
print("DataFrame with NaNs:\n", df_missing)

# Drop rows with NaNs
print("\nDrop NaNs (dropna):\n", df_missing.dropna())

# Fill NaNs with a specific value
print("\nFill NaNs with 0 (fillna):\n", df_missing.fillna(0))
```

## 8. Grouping and Aggregation (Split-Apply-Combine)

```python
# Grouping data
df_group = pd.DataFrame({
    'Department': ['IT', 'HR', 'IT', 'Sales', 'HR', 'Sales'],
    'Salary': [60000, 55000, 65000, 70000, 52000, 72000]
})

# Group by department and calculate mean salary
grouped = df_group.groupby('Department')['Salary'].mean()
print("Mean salary by Department:\n", grouped)
```

## 9. Merging, Joining, and Concatenating

```python
df_a = pd.DataFrame({'ID': [1, 2, 3], 'Value_A': ['A1', 'A2', 'A3']})
df_b = pd.DataFrame({'ID': [2, 3, 4], 'Value_B': ['B2', 'B3', 'B4']})

# Merge (Inner join by default)
merged_inner = pd.merge(df_a, df_b, on='ID')
print("Inner Merge on ID:\n", merged_inner)
```

## 10. Advanced Operations: Apply and Lambda Functions
Applying custom functions to DataFrames.

```python
# Using apply to modify a column
df_apply = pd.DataFrame({'Prices': [100, 200, 300]})

# Using lambda functions for quick inline operations
df_apply['Taxed'] = df_apply['Prices'].apply(lambda x: x * 1.05)
print("\nApplying a lambda function:\n", df_apply)
```
