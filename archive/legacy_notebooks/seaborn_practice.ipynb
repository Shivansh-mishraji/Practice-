# Seaborn Practice

Seaborn is a Python data visualization library based on Matplotlib. It provides a high-level interface for drawing attractive and informative statistical graphics. It integrates closely with Pandas DataFrames.

## 1. Importing Seaborn

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

# Set seaborn style
sns.set_theme(style="darkgrid")
print("Seaborn version:", sns.__version__)
```

## 2. Loading Built-in Datasets

```python
# Load the 'tips' dataset
tips = sns.load_dataset('tips')
print("Tips Dataset Head:\n", tips.head())
```

## 3. Relational Plots (relplot)

```python
sns.relplot(
    data=tips,
    x="total_bill", y="tip", col="time",
    hue="smoker", style="smoker", size="size"
)
plt.show()
```

## 4. Categorical Plots (catplot)

```python
# Swarm Plot
sns.catplot(data=tips, x="day", y="total_bill", hue="sex", kind="swarm")
plt.show()

# Box Plot
sns.catplot(data=tips, x="day", y="total_bill", hue="smoker", kind="box")
plt.show()
```

## 5. Matrix Plots (Heatmaps)

```python
numeric_tips = tips.select_dtypes(include=[np.number])
corr_matrix = numeric_tips.corr()

plt.figure(figsize=(6, 4))
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm", center=0)
plt.title("Correlation Heatmap")
plt.show()
```

## 6. Joint Plots (jointplot)

```python
# Scatter plot with marginal histograms
sns.jointplot(data=tips, x="total_bill", y="tip", hue="time")
plt.show()
```

## 7. Linear Regression Models (lmplot)

```python
# Scatter plot with a linear regression line
sns.lmplot(data=tips, x="total_bill", y="tip", hue="smoker", col="time")
plt.show()
```
