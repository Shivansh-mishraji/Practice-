# Matplotlib Practice

Matplotlib is the foundational plotting library for Python. It provides an object-oriented API for embedding plots into applications, and its `pyplot` module offers a MATLAB-like interface.

## 1. Importing Matplotlib

```python
import matplotlib.pyplot as plt
import numpy as np

# Setting a style for better looking plots
plt.style.use('ggplot') 
```

## 2. Basic Plotting: Line Plot

```python
# Generate data
x = np.linspace(0, 10, 100)
y = np.sin(x)

# Create a figure and axis explicitly
fig, ax = plt.subplots(figsize=(8, 4)) 

# Plot data
ax.plot(x, y, label='sin(x)', color='blue', linestyle='--', linewidth=2)

# Add title and labels
ax.set_title('Basic Line Plot: Sine Wave')
ax.set_xlabel('X Axis')
ax.set_ylabel('Y Axis')

# Add a legend
ax.legend()
plt.show()
```

## 3. Multiple Plots on Same Axis

```python
y2 = np.cos(x)

fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(x, y, label='sin(x)', color='blue')
ax.plot(x, y2, label='cos(x)', color='red', linestyle='-.')

ax.set_title('Multiple Lines')
ax.legend()
plt.show()
```

## 4. Scatter Plots

```python
np.random.seed(42)
x_scatter = np.random.rand(50)
y_scatter = np.random.rand(50)
colors = np.random.rand(50)
sizes = 1000 * np.random.rand(50)

fig, ax = plt.subplots(figsize=(8, 5))
scatter = ax.scatter(x_scatter, y_scatter, c=colors, s=sizes, alpha=0.5, cmap='viridis')

cbar = fig.colorbar(scatter)
cbar.set_label('Color Intensity')

ax.set_title('Scatter Plot')
plt.show()
```

## 5. Bar Charts

```python
categories = ['Apples', 'Bananas', 'Cherries', 'Dates']
values = [25, 40, 15, 30]

fig, ax = plt.subplots(figsize=(8, 4))
ax.bar(categories, values, color=['red', 'yellow', 'darkred', 'brown'])

ax.set_title('Fruit Sales')
ax.set_ylabel('Quantity Sold')
plt.show()
```

## 6. Subplots

```python
# Create a 2x2 grid of plots
fig, axes = plt.subplots(nrows=2, ncols=2, figsize=(10, 8))

# Access each subplot using indexing
axes[0, 0].plot(x, y, color='blue')
axes[0, 0].set_title('Sine')

axes[0, 1].plot(x, y2, color='red')
axes[0, 1].set_title('Cosine')

axes[1, 0].scatter(x_scatter, y_scatter, alpha=0.5)
axes[1, 0].set_title('Scatter')

axes[1, 1].bar(categories, values)
axes[1, 1].set_title('Bar')

plt.tight_layout()
plt.show()
```
